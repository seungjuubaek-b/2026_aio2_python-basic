```python
def local_example():
    local_var = "이것은 지역 변수입니다"
    print(local_var)
local_example()

global_var = "이것은 전역 변수입니다"
def global_example():
    print(global_var)
global_example()

class Example:
    static_var = "이것은 정적 변수입니다"
    def print_static():
        print(Example.static_var)
print(Example.static_var)
```
