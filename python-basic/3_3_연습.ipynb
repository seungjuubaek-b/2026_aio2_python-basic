```python
# 클래스 정의
class Student :
    def __init__(self, name, age) :
        self.name = name + '@'
        self.age = age
    def introduce(self) :
        print(f"안녕하세요 제 이름은 {self.name} 입니다. {self.age}살입니다.")
```

```python
s1 = Student("alice", 25)
s1.introduce()
s2 = Student("billy", 35)
s2.introduce()
```

```python
class Calculator :
    def __init__(self, number_list) :
        self.number_list = number_list
    def sum(self) :
        return sum(self.number_list)
    def avg(self) :
        return sum(self.number_list) / len(self.number_list)
```

```python
cal1 = Calculator([1,2,3,4,5])
print(cal1.sum(), cal1.avg())
cal2 = Calculator([6,7,8,9,10])
print(cal2.sum(), cal2.avg())
```

```python
import random

class Account :
    bank_name = 'SC은행' #클래스 변수

    def __init__(self, name, balance) :
        self.name = name # 인스턴스 변수
        self.balance = balance
        self.account_no = Account.make_account_no()

    @staticmethod
    def make_account_no() :
        num1 = random.randint(0,999)
        num2 = random.randint(0,99)
        num3 = random.randint(0,999999)

        return f"{num1:03d}-{num2:02d}-{num3:06d}"

    def print_info(self) :
        return f'은행이름 : {Account.bank_name} \n 계좌번호 : {self.account_no}'

    def 입금(self, amount) :
        self.balance += amount

    def 출금(self, amount) :
        self.balance -= amount

    def 잔고조회(self) :
        return f'{self.name}님의 잔고는 {self.balance}원입니다.'
```

```python
ac1 = Account('홍길동', 2000)
ac2 = Account('손흥민', 3000)
print(ac1.print_info())
ac1.잔고조회()
ac1.make_account_no()

print(ac2.print_info())
ac2.잔고조회()
ac2.make_account_no()
```

```python
class Animal :
    def make_sound(self) :
        print("어흥")
class Dog(Animal) :
    def make_sound(self) :
        return "멍멍!"
class Cat(Animal) :
    def make_sound(self):
        return "야옹!"
```

```python
Dog().make_sound()
```

```python
Cat().make_sound()
```

```python
animal = [Dog(), Cat(), Dog(), Cat()]
for ani in animal :
    print(ani.make_sound())
```
