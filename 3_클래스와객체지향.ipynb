- 사용자정의 자료구조 : 클래스

    2) 행위 / 동작 : class/static/self

객체지향으로 프로그램을 짜줘

```python
#클래스 정의
class Student :
    def __init__(self, name, age) :
        self.name = name + '@'
        self.age = age
    def introduce(self) :
        print(f"안녕하세요 제 이름은 {self.name} 입니다. {self.age}살입니다.")
```

# 객체 생성 & 호출

```python
s1 = Student("alice", 25)
s1.introduce() # 메소드
s2 = Student("billy", 35)
s2.introduce()
```

## class Calculator 정의

```python
#클래스 정의
class Calculator :
    def __init__(self, number_list) :
        self.number_list = number_list
    def sum(self) :
        return sum(self.number_list)
    def avg(self) :
        return sum(self.number_list)/ len(self.number_list)
```

```python
# Q1. 다음과 같이 동작하는 클래스 Calculator를 작성해 보자.
cal1 = Calculator( [1,2,3,4,5] )
print(cal1.sum(), cal1.avg()) # 15

cal2 = Calculator([6,7,8,9,10])
print(cal2.sum(), cal2.avg()) # 40
```

## Account 클래스 정의

```python
import random


class Account:
    bank_name = 'SC은행' #클래스 변수
   
    def __init__(self, name, balance):
        self.name = name #인스턴스 변수
        self.balance = balance
        self.account_no = Account.make_account_no()
   
    @staticmethod    
    def make_account_no():
        num1 = random.randint(0, 999)
        num2 = random.randint(0, 99)
        num3 = random.randint(0, 999999)


        return f"{num1:03d}-{num2:02d}-{num3:06d}"
   
    def print_info(self,):
        return f'은행이름 : {Account.bank_name} \n계좌번호: {self.account_no}'
   
    def 입금(self, amount):
        self.balance += amount
   
    def 출금(self, amount):
        self.balance -= amount
   
    def 잔고조회(self):
        return f' {self.name}님의 잔고는 {self.balance} 입니다'
```

## 객체 생성

```python
ac1 = Account('홍길동', 2000)
ac2 = Account('손흥민', 3000)
print(ac1.print_info())
ac1.잔고조회()
ac1.make_account_no()
```

# 상속과 재정의(Overriding)

```python
class Animal:
    def make_sound(self):
        print("어흥")
class Dog(Animal): # default로 __init__() 함수가 정의되어 있음!
    # pass
    def make_sound(self):
        return "멍멍!"
class Cat(Animal):
    # pass
    def make_sound(self):
        return "야옹!"
```

```python
Dog().make_sound()
```

```python
Cat().make_sound()
```

## 객체의 연산자

```python
animal = [Dog(), Cat(), Dog(), Cat()]
for ani in animal :
    print(ani.make_sound())
```

# 예외 클래스 정의

```python
class AgeException(Exception) :
    def __init__(self, msg) :
        self.message = msg
```

```python
def input_age() :
    # 나이를 입력
    age = int(input("나이를 입력하세요: "))
    # 입력값 체크
    if age < 0 : # 나이는 양수
        raise AgeException("나이는 양수입니다.")
    elif age > 150 : # 상한 체크
        raise AgeException("진짜?")
    else :
        return age
```

```python
# 예외처리를 하는 나이 입력 구현
try :
    age = input_age()

except AgeException as e :
    print(e.args[0])

else :
    print(f"입력하신 나이는 {age}입니다.")
```
