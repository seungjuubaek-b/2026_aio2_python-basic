# 연산자

`123`은 숫자형(`int`)이고 `"123"`은 문자열형(`str`)입니다.  
Python에서는 값이 비슷해 보여도 자료형이 다르면 다른 값으로 비교됩니다.  
따라서 `123 == "123"`의 결과는 `False`입니다.

```python
123 == "123"
```

```python
123 == 123 # == '같다'라는 연산자
```
